```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.preprocessing import normalize
from sklearn.cluster import AgglomerativeClustering
```

```python
df = pd.read_csv('clientes.csv', delimiter=";")
```

```python
print(df.head())
```

Creamos un dedrograma para deducir el número de clústers

```python
dendrogram(linkage (df, method = 'ward', metric = 'euclidean'))
```

Agrupamiento jerárquico aglomerativo

```python
agrup = AgglomerativeClustering (n_clusters=3,linkage='ward')
agrup.fit(df)
print(agrup.labels_)
```

```python
plt.scatter(df['Salario'],df['Gasto'], c=agrup.labels_)
plt.xlabel('Salario Mensual')
plt.ylabel('Gasto (%)')
```
