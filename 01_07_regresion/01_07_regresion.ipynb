```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
regresion_lineal = LinearRegression()
```

```python
df = pd.read_csv('empleados.csv', delimiter=";")
```

```python
print(df.head(20))
```

```python
print(df.isnull().sum())
```

Separación de datos

```python
data_null = df[df['Edad'].isnull()==True]
train_set = df[df['Edad'].isnull()== False]
```

Diferenciamos entre labels (y: los datos de edad) empleadas en el entrenamiento y samples (x:todo el resto de datos para cada fila) que utilizamos para entrenar el modelo de regresión.

```python
x_train = train_set.drop('Edad',axis=1)
y_train = train_set['Edad']
print(x_train.shape)
print(y_train.shape)
```

```python
regresion_lineal.fit(x_train,y_train)
```

```python
data_null = data_null.drop('Edad', axis=1)
pred = regresion_lineal.predict(data_null)
print (pred)
```

```python
data_null['Edad'] = pred.round()
print(data_null)
```

Sustituímos los datos obtenidos en el df original para tener todos los datos completos.

```python
df['Edad'] = df['Edad'].fillna(data_null['Edad'])
print (df.head(20))
```
