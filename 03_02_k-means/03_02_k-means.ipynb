```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df = pd.read_csv('clusters.csv', delimiter=";")
```

```python
print(df.head(10))
```

Dividir en 4 grupos usando KMeans

```python
kmeans = KMeans(n_clusters=4)
kmeans.fit(df)
y = kmeans.labels_
centers = kmeans.cluster_centers_
```

Representación

```python
plt.scatter(df['Latitud'],df['Longitud'], c=y, s=20)
plt.scatter(centers[:,0],centers[:,1], c='red')
plt.xlabel('Latitud')
plt.ylabel('Longitud')
```
